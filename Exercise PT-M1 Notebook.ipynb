### Cell 3: Loading Custom Data & Chunking
* **Code Explanation:**
  - `DirectoryLoader(...)`: This initialization creates a LangChain document loader specifically configured to scan an entire directory rather than a single file. We pass 'my_data/' as the target directory and use the glob parameter '**/*.pdf' to instruct the loader to recursively find all files ending with the PDF extension. We also specify `PyPDFLoader` as the underlying extraction engine to handle the parsing of these specific binary files.
  - `raw_documents = loader.load()`: This command executes the actual physical loading process, triggering the `DirectoryLoader` to iterate through all the discovered PDF files in the target directory. As it processes the files, it extracts the raw text from each individual page and converts it into a standardized LangChain `Document` object. These generated objects contain both the extracted text payload and vital metadata, such as the original source file name and the specific page number.
  - `RecursiveCharacterTextSplitter(...)`: This instantiates a sophisticated text splitting algorithm designed to break our massive, multi-page document objects into much smaller, semantically meaningful segments. The `chunk_size=700` parameter dictates the target maximum length for each individual chunk, ensuring they are small enough to fit comfortably within the LLM's context window. The `chunk_overlap=70` parameter forces the algorithm to duplicate the last 70 characters of a previous chunk at the beginning of the next one, which prevents sentences or core concepts from being abruptly severed across chunk boundaries.
  - `documents = text_splitter.split_documents(raw_documents)`: This line applies the previously configured recursive splitting logic directly to our array of large, raw PDF document objects. It iterates through the text of every single page, slicing the content according to our size and overlap rules, and generates a massive new array of smaller, bite-sized chunk objects. These resulting chunks are now fully optimized for the embedding process and subsequent storage within our vector database.
* **Architectural Necessity:** LLMs suffer from context window limitations and lost in the middle phenomena. Furthermore, sending entire manuals to an LLM for every single query is highly inefficient and expensive. By chunking the text, the RAG system can isolate and retrieve only the specific paragraphs that contain the answer to the user's query. The overlap ensures that a sentence or concept is not awkwardly cut in half between two chunks, preserving semantic context.

```python
# Cell 3: Load and split the PDF manuals
import os
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = DirectoryLoader("my_data/", glob="**/*.pdf", loader_cls=PyPDFLoader, show_progress=True)
raw_documents = loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=70
)
documents = text_splitter.split_documents(raw_documents)

print(f"Loaded {len(raw_documents)} pages across all manuals and split into {len(documents)} chunks.")
```

### Cell 4: Embedding Model & Vector DB Indexing
* **Code Explanation:**
  - `HuggingFaceEmbeddings(...)`: This command initializes a highly specialized, open-source embedding model that will run locally on our machine. We specifically selected the `all-MiniLM-L6-v2` model because it is incredibly lightweight, fast, and optimized for general-purpose semantic similarity tasks. Its primary job is to take human-readable text strings and translate them into 384-dimensional dense numerical vectors that capture the underlying semantic meaning of the words.
  - `Chroma.from_documents(...)`: This function serves as the central hub for our data ingestion pipeline, performing two massive, resource-intensive operations simultaneously. First, it iterates through our array of chunked documents and passes each one through the HuggingFace embedding model to generate a corresponding mathematical vector. Second, it instantiates a brand new, in-memory Chroma vector database and permanently stores both the original text chunks and their newly calculated vector representations for future retrieval.
  - `retriever = vectorstore.as_retriever(...)`: This line takes our newly populated Chroma vector database and wraps it in a LangChain retriever interface, transforming it from a static storage system into an active search engine. By passing the `search_kwargs={"k": 3}` argument, we strictly configure the retriever to only return the top 3 most mathematically similar document chunks when presented with a query. Limiting the return size prevents us from overwhelming the LLM's context window with excessive or irrelevant data during the final generation phase.
* **Architectural Necessity:** LLMs understand mathematics, not raw text. Embeddings capture the semantic meaning of the text. When a user asks a question, their query is also converted into an embedding. The vector database then performs a cosine similarity search, rapidly calculating which stored chunks are mathematically closest to the user's question. This retrieval step provides the external knowledge that grounds the final answer.

```python
# ==============================================================================
# Cell 4: Embedding Model & Vector DB Indexing
# ==============================================================================
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

# Initialize open-source embedding model
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Store embeddings into Chroma vector database
vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings
)

# Set vectorstore as a retriever
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
```

### Cell 5: Model Initialization and Domain System Prompt
* **Code Explanation:**
  - `ChatGroq(...)`: This instantiates our core language model connection using the highly optimized LangChain Groq integration. We specify the exact model name we wish to use and, critically, we set the `temperature` parameter explicitly to `0`. Setting the temperature to zero eliminates the model's creative randomness, forcing it to behave in a highly deterministic, factual, and predictable manner which is absolutely essential for a reliable RAG assistant.
  - `system_prompt`: This massive string variable establishes the overarching behavioral guidelines, strict operational boundaries, and assigned persona for our LLM assistant. It explicitly commands the AI to completely ignore its vast pre-trained knowledge base and formulate answers using *only* the context provided by our vector database. Furthermore, it dictates a precise, hardcoded fallback phrase ('I cannot answer based on the provided domain data') that the model must output if the retrieved context does not contain the answer, acting as a primary defense against hallucination.
  - `ChatPromptTemplate.from_messages(...)`: This function takes our static, plain-text system instructions and combines them with a placeholder for the dynamic user input to construct a formal LangChain prompt template. This template acts as the architectural blueprint for every single interaction, guaranteeing that our strict system prompt is seamlessly injected into the conversation flow before the user's query. By using this structured template format, LangChain can systematically parse the context, the guidelines, and the question, formatting them perfectly for the Groq API.
* **Architectural Necessity:** Even with relevant context retrieved, LLMs have a tendency to rely on their pre-trained weights and hallucinate. The strict system prompt acts as a governance layer, forcing the model to act solely as a synthesizer of the retrieved context rather than an independent oracle. This is essential for enterprise or domain-specific RAG where accuracy is paramount.

```python
# # Cell 5: Model Initialization and Domain System Prompt
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

# Initialize the model with an available active model ID
llm = ChatGroq(
    model_name="openai/gpt-oss-20b",
    temperature=0
)

# Custom domain system prompt
system_prompt = (
    "You are a specialized AI assistant for the user's uploaded domain.\n"
    "Answer questions strictly using ONLY the provided context below.\n"
    "If the answer cannot be found in the context, reply: 'I cannot answer based on the provided domain data.'\n\n"
    "Context:\n{context}"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])
```

### Cell 6: Pipeline Assembly
* **Code Explanation:**
  - `def format_docs(docs):`: This is a custom Python utility function designed to handle the final text processing step before the context is fed to the LLM. It accepts a list of isolated LangChain `Document` objects returned by our vector database and extracts the raw text payload from each one. It then joins these extracted payloads together using double newlines, transforming the fragmented chunks into one single, highly readable block of cohesive text.
  - `class RagPipeline:`: By defining this custom Python class, we are encapsulating the entire, complex Retrieval-Augmented Generation workflow into a single, easily manageable object. This object-oriented approach allows us to tightly bind our distinct architectural components�the database retriever, the prompt template, and the LLM engine�into a unified system. This makes the code much more modular, scalable, and easier to debug, as the entire pipeline can be instantiated and executed cleanly.
  - `def invoke(self, inputs):`: This method serves as the primary execution engine for our custom RAG class, orchestrating the precise sequence of events required to answer a user's question. First, it extracts the user's query string and immediately feeds it into the vector database retriever to surface the three most relevant document chunks. Next, it passes those raw chunks through our `format_docs` helper function to stitch them together into a single context block. Finally, it uses LangChain Expression Language (LCEL) piping syntax to inject both the context and the query into the prompt template, send the completed prompt to the Groq LLM, and return the final generated answer alongside the source citations.
* **Architectural Necessity:** This cell represents the Augmented Generation portion of RAG. It handles the critical handoff between the retrieval database and the generation model. By packaging this logic into a unified method, the complex underlying mechanics are abstracted away, leaving a clean interface for the end-user application to submit questions and receive answers.

```python
# # Cell 6: Pipeline Assembly
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Format retrieved documents into a single text block
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# Create an equivalent RAG pipeline using LangChain Expression Language
combine_docs_chain = prompt | llm | StrOutputParser()

class RagPipeline:
    def __init__(self, retriever, prompt, llm):
        self.retriever = retriever
        self.prompt = prompt
        self.llm = llm

    def invoke(self, inputs):
        query = inputs["input"]
        docs = self.retriever.invoke(query)
        formatted_context = format_docs(docs)
        response_text = (self.prompt | self.llm | StrOutputParser()).invoke({
            "context": formatted_context,
            "input": query
        })
        return {"answer": response_text, "context": docs}

rag_chain = RagPipeline(retriever, prompt, llm)
```

### Cell 7: Testing Your Custom Domain Chatbot (In-Domain Query)
* **Code Explanation:**
  - `user_query = ...`: This line initializes a standard string variable containing a specific, targeted test question designed to query the core capabilities of our system. We deliberately chose a question that asks for a summary of key information, which forces the system to pull data from multiple different chunks across the uploaded hardware manuals. This ensures we are testing the system's ability to synthesize information, rather than just performing a simple keyword lookup.
  - `response = rag_chain.invoke(...)`: This executes the core `invoke` method of our custom RAG pipeline class, passing in the user query we defined in the previous step. Under the hood, this single line triggers the entire complex chain of events: generating an embedding for the query, querying the ChromaDB for nearest neighbors, formatting the prompt, and calling the Groq API. The resulting output is a comprehensive dictionary containing both the final LLM-generated string and the raw source document objects.
  - `print(...)`: These standard Python print statements are utilized to output the final results of the RAG pipeline directly to the Jupyter Notebook console for visual inspection. We clearly demarcate the section by printing a header, followed immediately by the actual text response generated by the LLM. This provides immediate, human-readable feedback to the developer, confirming whether the system correctly understood the prompt and synthesized a factual answer.
  - `for i, doc in enumerate(response["context"]):`: This block establishes a loop that systematically iterates through the raw array of source document chunks that our pipeline returned alongside the generated answer. During each iteration, it digs into the specific chunk's metadata dictionary, extracts the original file name, and prints it to the console. This step is absolutely critical for evaluating a RAG system, as it provides total transparency and 'explainability,' allowing the developer to manually verify that the LLM actually pulled its facts from the correct source files.
* **Architectural Necessity:** This step is vital for evaluating the baseline performance of the RAG system. It proves that the entire end-to-end architecture is functioning correctly. Furthermore, printing the source chunks provides explainability, allowing developers and users to verify exactly where the LLM pulled its facts from.

```python
# ==============================================================================
# Cell 7: Testing Your Custom Domain Chatbot (In-Domain Query)
# ==============================================================================
# Test Case 1: In-Domain Query
user_query = "Summarize the key information found in the documents."
response = rag_chain.invoke({"input": user_query})

print("--- DOMAIN QUERY ANSWER ---")
print(response["answer"])

print("\n--- RETRIEVED SOURCE CHUNKS ---")
for i, doc in enumerate(response["context"]):
    print(f"Chunk {i+1} Source:", doc.metadata.get("source", "Unknown"))
```

### Cell 8: Experimental Tasks (Fallback & Hallucination Stress-Test)
* **Code Explanation (Task 2: Fallback Test):**
  - `ood_query = ...`: Here, we deliberately define a question about baking sourdough bread, which is entirely outside the scope of our computer hardware manuals. The purpose of this out-of-domain query is to intentionally trigger a failure state in our retrieval mechanism. Because the vector database contains no information about baking, the retrieved chunks will be completely irrelevant to the question.
  - `ood_response = rag_chain.invoke(...)`: We pass this irrelevant sourdough query into our strictly configured RAG pipeline to test the robustness of our safety guardrails. When the LLM receives the prompt, it will analyze the irrelevant context chunks retrieved by the database and realize it cannot answer the question. It will then follow the strict instructions in our system prompt, triggering the fallback mechanism and returning the precise error phrase we hardcoded earlier.

* **Code Explanation (Task 3: Hallucination Stress-Test):**
  - `stress_llm = ChatGroq(..., temperature=1.0)`: To conduct our hallucination stress test, we deliberately initialize a brand new connection to the Groq LLM with drastically altered configuration parameters. Most importantly, we crank the `temperature` parameter all the way up to its maximum value of 1.0. This maximizes the model's creative randomness and significantly increases the probability that it will ignore constraints, take extreme liberties with facts, and generate hallucinated content.
  - `stress_system_prompt = ...`: In this step, we completely strip away all of the protective guardrails and governance mechanisms we carefully built into our previous system prompt. We completely remove the critical instruction forcing the model to 'only use the provided context,' and we completely delete the hardcoded fallback phrase. This leaves the model completely unconstrained, effectively granting it permission to rely on its vast, pre-trained knowledge base to answer questions.
  - `stress_response = stress_llm.invoke(...)`: For the final phase of the stress test, we completely bypass our custom RAG pipeline and directly feed the out-of-domain sourdough query into this highly creative, completely unconstrained LLM instance. Because the model has a high temperature and lacks strict grounding instructions, it will entirely ignore its assigned persona as a computer assistant. Instead, it will hallucinate a detailed, confident, and entirely ungrounded recipe for sourdough bread, concretely demonstrating why strict system prompts and zero temperature are absolute prerequisites for safe, production-grade RAG applications.
* **Architectural Necessity:** In LLMOps, you must evaluate how systems fail, not just how they succeed. Task 2 validates that our guardrails successfully prevent off-topic interactions. Task 3 acts as an ablation study, proving that without strict temperature control and rigid system prompts, the LLM will hallucinate and deviate from its intended domain.

```python
# ==============================================================================
# EXPERIMENTAL TASKS (Phase 4 of the Lab Exercise)
# ==============================================================================

# Task 2: Out-of-Domain Fallback Test
ood_query = "What are the primary ingredients needed to bake a sourdough loaf?"
ood_response = rag_chain.invoke({"input": ood_query})

print("--- TASK 2 FALLBACK TEST ---")
print("Query:", ood_query)
print("Response:", ood_response["answer"])


# Task 3: Hallucination Stress-Test (Direct Execution)
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

# 1. Elevate temperature to 1.0
stress_llm = ChatGroq(
    model_name="openai/gpt-oss-20b",
    temperature=1.0
)

# 2. Strip out the fallback restriction
stress_system_prompt = (
    "You are an AI assistant for computer troubleshooting.\n\n"
    "Context:\n{context}"
)

stress_prompt = ChatPromptTemplate.from_messages([
    ("system", stress_system_prompt),
    ("human", "{input}"),
])

# 3. Retrieve context and run directly without langchain.chains
retrieved_docs = retriever.invoke(ood_query)
context_text = "\n\n".join([doc.page_content for doc in retrieved_docs])

formatted_prompt = stress_prompt.format_messages(
    context=context_text,
    input=ood_query
)
stress_response = stress_llm.invoke(formatted_prompt)

print("--- TASK 3 HALLUCINATION STRESS-TEST ---")
print("Query:", ood_query)
print("Response (Ungrounded / Temp 1.0):\n", stress_response.content)
```

### Cell 9: Interactive Chatbot Interface
* **Code Explanation:**
  - `print(...)`: We start by printing a welcoming message and instructions on how to exit the interactive loop so the user knows the chatbot is active. This provides a user-friendly entry point for the session.
  - `while True:`: This creates an infinite loop that will keep the chatbot running continuously. It allows for multiple back-and-forth interactions without needing to constantly re-run the code cell manually.
  - `user_input = input("You: ")`: The `input()` function is a built-in Python command that pauses the loop and generates a text box in the Jupyter interface. It actively waits for the user to type a custom question and press enter.
  - `if user_input.lower() in ['exit', 'quit']:`: This conditional acts as the critical escape hatch for our infinite loop. If the user types 'exit' or 'quit', the `break` command is triggered, safely terminating the background process and ending the chatbot session.
  - `response = rag_chain.invoke(...)`: Just like in our static tests, this passes the dynamically captured user input directly into our custom RAG pipeline. The pipeline then handles the embedding, retrieval, and generation steps seamlessly in the background.
  - `print(...)` and `for ... in ...:`: Finally, this cleanly formats the LLM's generated answer and iterates through the source chunks. It prints the citations so the user can immediately verify exactly which manual the answer came from.
* **Architectural Necessity:** A true chatbot needs to be interactive rather than hardcoded. This cell transforms our static RAG pipeline testing script into a dynamic, usable application. By wrapping the `invoke` method in a continuous input loop, we simulate a real-world user interface where domain queries can be submitted on the fly and evaluated in real-time.

```python
# ==============================================================================
# Cell 9: Interactive Chatbot Interface
# ==============================================================================
print("?? Custom Domain RAG Chatbot is Ready!")
print("Type 'exit' or 'quit' to stop.\n")

while True:
    user_input = input("You: ")
    if user_input.lower() in ['exit', 'quit']:
        print("Chatbot session ended.")
        break
        
    if not user_input.strip():
        continue
        
    print("\nProcessing...")
    try:
        response = rag_chain.invoke({"input": user_input})
        print(f"\n?? Chatbot: {response['answer']}")
        print("\n[Source Documents Used:]")
        for i, doc in enumerate(response['context']):
            print(f"- {doc.metadata.get('source', 'Unknown')}")
    except Exception as e:
        print(f"\n?? Error occurred: {e}")
        
    print("-" * 50 + "\n")
```
